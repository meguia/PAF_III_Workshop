# Series de Fourier, parte II

Los coeficientes de Fourier se obtienen promediando la senal contra el conjugado de cada oscilacion elemental. Para la onda cuadrada solo aparecen armonicos impares.


```python
import numpy as np
import matplotlib.pyplot as plt

try:
    from ipywidgets import interact, FloatSlider, IntSlider, Checkbox, Dropdown
    WIDGETS_AVAILABLE = True
except Exception:
    WIDGETS_AVAILABLE = False

plt.rcParams["figure.figsize"] = (8, 4.5)
plt.rcParams["axes.grid"] = True
```

```python
def square_series(N=9):
    x = np.linspace(0, 2*np.pi, 1200)
    y = np.zeros_like(x)
    for k in range(1, N+1, 2):
        y += (4/(np.pi*k))*np.sin(k*x)
    target = np.where(x < np.pi, 1, -1)
    fig, ax = plt.subplots(figsize=(10, 4))
    ax.plot(x, target, color="0.3", lw=2, label="cuadrada ideal")
    ax.plot(x, y, color="C3", label=f"suma hasta k={N}")
    ax.set_ylim(-1.6, 1.6)
    ax.legend()
    plt.show()

if WIDGETS_AVAILABLE:
    interact(square_series, N=IntSlider(min=1, max=41, step=2, value=9))
else:
    square_series()
```
