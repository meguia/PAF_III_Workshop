# Series de Fourier, parte II

La DFT transforma un array de samples en un array de coeficientes. Para `N` samples:

$C_k=\frac{1}{N}\sum_{n=0}^{N-1}s[n]e^{-i2\pi kn/N}$.

Y la reconstruccion es:

$s[n]=\sum_{k=0}^{N-1}C_k e^{i2\pi kn/N}$.

Todo se calcula con sumas sobre samples discretos.


```python
import numpy as np
import matplotlib.pyplot as plt

try:
    from ipywidgets import interact, FloatSlider, IntSlider, Checkbox, Dropdown
    WIDGETS_AVAILABLE = True
except Exception:
    WIDGETS_AVAILABLE = False

plt.rcParams["figure.figsize"] = (8, 4.5)
plt.rcParams["axes.grid"] = True
```

```python
def square_dft(N=128, Kmax=9):
    n = np.arange(N)
    s = np.where(n < N/2, 1.0, -1.0)
    C = np.fft.fft(s) / N

    recon = np.zeros(N, dtype=complex)
    bins = [0]
    for k in range(1, Kmax + 1):
        bins.extend([k, N-k])
    for k in bins:
        recon += C[k] * np.exp(1j * 2*np.pi*k*n/N)

    fig, (ax0, ax1) = plt.subplots(1, 2, figsize=(11, 4))
    ax0.stem(n, s, basefmt=" ", linefmt="0.6", markerfmt="o", label="samples")
    ax0.plot(n, recon.real, color="C3", label="reconstruccion parcial")
    ax0.set_title("senal sampleada")
    ax0.legend()

    freqs = np.fft.fftfreq(N) * N
    ax1.stem(freqs[:N//2], np.abs(C[:N//2]), basefmt=" ")
    ax1.set_xlim(0, min(N//2, Kmax + 8))
    ax1.set_title("|C_k|")
    ax1.set_xlabel("bin k")
    plt.show()

if WIDGETS_AVAILABLE:
    interact(square_dft, N=IntSlider(min=32, max=256, step=32, value=128),
             Kmax=IntSlider(min=1, max=31, step=2, value=9))
else:
    square_dft()
```
