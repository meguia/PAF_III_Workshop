# Derivada

Version en Python/NumPy del notebook de Pluto. La interactividad de Pluto se aproxima con `ipywidgets` cuando estan disponibles; si no, las funciones se pueden llamar manualmente.

La derivada es la pendiente local. Para una funcion $f$, la derivada en $x_0$ dice como cambia $f(x)$ ante un cambio pequeno de $x$.


```python
import numpy as np
import matplotlib.pyplot as plt

try:
    from ipywidgets import interact, FloatSlider, IntSlider, Checkbox, Dropdown
    WIDGETS_AVAILABLE = True
except Exception:
    WIDGETS_AVAILABLE = False

plt.rcParams["figure.figsize"] = (8, 4.5)
plt.rcParams["axes.grid"] = True
```

```python
def poly_and_derivative(a0=0, a1=1, a2=0, a3=0, a4=0, a5=0, x0=1.0, order=1, tangent=True):
    coeffs = np.array([a5, a4, a3, a2, a1, a0], dtype=float)
    x = np.linspace(-3, 3, 500)
    y = np.polyval(coeffs, x)
    dcoeffs = np.polyder(coeffs, m=order)
    yd = np.polyval(dcoeffs, x)
    y0 = np.polyval(coeffs, x0)
    slope = np.polyval(np.polyder(coeffs), x0)
    fig, ax = plt.subplots()
    ax.plot(x, y, label="f(x)")
    if order > 0:
        ax.plot(x, yd, label=f"derivada orden {order}")
    ax.scatter([x0], [y0], color="C3", zorder=5)
    if tangent:
        ax.plot(x, y0 + slope*(x-x0), "--", label="tangente en x0")
    ax.set_title("Polinomio y derivadas")
    ax.legend()
    plt.show()

if WIDGETS_AVAILABLE:
    interact(poly_and_derivative,
             a0=(-2.0, 2.0, 0.1), a1=(-2.0, 2.0, 0.1), a2=(-1.0, 1.0, 0.05),
             a3=(-1.0, 1.0, 0.05), a4=(-0.2, 0.2, 0.01), a5=(-0.1, 0.1, 0.005),
             x0=(-3.0, 3.0, 0.05), order=(0, 5), tangent=True)
else:
    poly_and_derivative()
```

## Funciones trigonometricas y exponenciales

Aqui se muestran derivadas numericas con diferencias finitas. Es una buena aproximacion visual, aunque para calculo simbolico exacto conviene usar SymPy u otra libreria simbolica.


```python
def finite_derivative(fun, x, n=1, h=1e-3):
    y = fun(x)
    for _ in range(n):
        y = np.gradient(y, x)
    return y

def trig_exp_derivative(kind="sin(ax)", a=1.0, x0=1.0, order=1):
    x = np.linspace(-4, 4, 1000)
    if kind == "sin(ax)":
        f = lambda u: np.sin(a*u)
    elif kind == "cos(ax)":
        f = lambda u: np.cos(a*u)
    else:
        f = lambda u: np.exp(a*u)
    y = f(x)
    yd = finite_derivative(f, x, order)
    fig, ax = plt.subplots()
    ax.plot(x, y, label="f(x)")
    ax.plot(x, yd, label=f"derivada orden {order}")
    ax.axvline(x0, color="0.3", ls="--")
    ax.legend()
    plt.show()

if WIDGETS_AVAILABLE:
    interact(trig_exp_derivative, kind=["sin(ax)", "cos(ax)", "exp(ax)"],
             a=(-1.0, 1.0, 0.1), x0=(-4.0, 4.0, 0.05), order=(0, 5))
else:
    trig_exp_derivative()
```
