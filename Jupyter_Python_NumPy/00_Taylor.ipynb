# Series de Taylor

Una expansion de Taylor aproxima una funcion cerca de un punto $a$ usando derivadas evaluadas en ese punto. La version de orden $N$ conserva informacion hasta la derivada $N$.


```python
import numpy as np
import matplotlib.pyplot as plt

try:
    from ipywidgets import interact, FloatSlider, IntSlider, Checkbox, Dropdown
    WIDGETS_AVAILABLE = True
except Exception:
    WIDGETS_AVAILABLE = False

plt.rcParams["figure.figsize"] = (8, 4.5)
plt.rcParams["axes.grid"] = True
```

```python
import math

def taylor_values(kind, x, a, N):
    if kind == "exp(x)":
        coeff = [np.exp(a) / math.factorial(n) for n in range(N+1)]
    elif kind == "sin(x)":
        derivs = [np.sin, np.cos, lambda u: -np.sin(u), lambda u: -np.cos(u)]
        coeff = [derivs[n % 4](a) / math.factorial(n) for n in range(N+1)]
    elif kind == "cos(x)":
        derivs = [np.cos, lambda u: -np.sin(u), lambda u: -np.cos(u), np.sin]
        coeff = [derivs[n % 4](a) / math.factorial(n) for n in range(N+1)]
    y = np.zeros_like(x, dtype=float)
    for n, c in enumerate(coeff):
        y += c * (x-a)**n
    return y

def taylor_demo(kind="sin(x)", a=0.0, N=3):
    x = np.linspace(-4, 4, 800)
    f = {"sin(x)": np.sin, "cos(x)": np.cos, "exp(x)": np.exp}[kind]
    fig, ax = plt.subplots()
    ax.plot(x, f(x), label=kind)
    ax.plot(x, taylor_values(kind, x, a, N), "--", label=f"Taylor orden {N}")
    ax.axvline(a, color="0.3", ls=":")
    ax.set_ylim(-4, 4)
    ax.legend()
    plt.show()

if WIDGETS_AVAILABLE:
    interact(taylor_demo, kind=["sin(x)", "cos(x)", "exp(x)"],
             a=(-3.0, 3.0, 0.05), N=(0, 8))
else:
    taylor_demo()
```
