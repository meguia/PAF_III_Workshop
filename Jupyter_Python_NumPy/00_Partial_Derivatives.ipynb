# Gradiente

El gradiente de $f(x,y)$ combina las derivadas parciales en las direcciones $x$ e $y$. Apunta hacia donde la funcion aumenta mas rapido.


```python
import numpy as np
import matplotlib.pyplot as plt

try:
    from ipywidgets import interact, FloatSlider, IntSlider, Checkbox, Dropdown
    WIDGETS_AVAILABLE = True
except Exception:
    WIDGETS_AVAILABLE = False

plt.rcParams["figure.figsize"] = (8, 4.5)
plt.rcParams["axes.grid"] = True
```

```python
def gradient_demo(kind="x^2+y^2", x0=1.0, y0=1.0, show_field=True, scale=1.0):
    x = np.linspace(-2, 2, 100)
    y = np.linspace(-2, 2, 100)
    X, Y = np.meshgrid(x, y)
    if kind == "x^2+y^2":
        Z = X**2 + Y**2
        Fx, Fy = 2*x0, 2*y0
        U, V = 2*X, 2*Y
    elif kind == "x^2-y^2":
        Z = X**2 - Y**2
        Fx, Fy = 2*x0, -2*y0
        U, V = 2*X, -2*Y
    else:
        Z = X*Y
        Fx, Fy = y0, x0
        U, V = Y, X
    fig, ax = plt.subplots(figsize=(6, 6))
    cs = ax.contourf(X, Y, Z, levels=30, cmap="coolwarm")
    plt.colorbar(cs, ax=ax)
    if show_field:
        step = 8
        ax.quiver(X[::step, ::step], Y[::step, ::step], U[::step, ::step], V[::step, ::step],
                  color="k", alpha=0.35)
    ax.scatter([x0], [y0], color="black")
    ax.quiver([x0], [y0], [Fx*scale], [Fy*scale], color="black", angles="xy", scale_units="xy", scale=1)
    ax.set_aspect("equal")
    ax.set_title(f"gradiente = ({Fx:.2f}, {Fy:.2f})")
    plt.show()

if WIDGETS_AVAILABLE:
    interact(gradient_demo, kind=["x^2+y^2", "x^2-y^2", "x*y"],
             x0=(-2.0, 2.0, 0.05), y0=(-2.0, 2.0, 0.05),
             show_field=True, scale=(0.05, 1.0, 0.05))
else:
    gradient_demo()
```
